## Read and process ELAN annotations

Now we have our data extracted from the videos. For training, we also need the ground truth so that we know in which time sequence ranges a movement took place and when not.

For this, the file must be annotated using a program such as ELAN, and the annotation then saved into a txt file.

It is important to note that in ELAN you should label the entire time range in which the movement occurs. The reason is that we only want, for example, the slideshow to change once, and not multiple times during the same movement. The task in the end is to return exactly one instance of the movement within this labeled range. This result is then either used to control the slideshow once or written into the events.csv file, which is then used to calculate the performance score.

Tip: Take a look at how we do it in the examples and follow the same approach. Make sure to always write rotate, swipe_left, and swipe_right exactly as we do in the examples.

Now the file can be read in and inspected:

```python
ELAN_ANNOTATION_FILE_PATH = "demo_data/demo_video_rotate_annotation.txt"
# read the ELAN file as CSV; only read the relevant columns (specified by `usecols`)
annotations = pd.read_csv(ELAN_ANNOTATION_FILE_PATH, sep="\t", header=None, usecols=[3, 5, 8], names=["start", "end", "label"])

# convert start and end timestamps into a datetime-datatype – this is important for the next step
annotations["start"] = pd.to_timedelta(annotations["start"], unit="s")
annotations["end"]   = pd.to_timedelta(annotations["end"], unit="s")

annotations
```

Since we now have the frames and also know when a movement occurred, we can add a column to the data called "ground_truth". In this column we write "idle" for all frames, except for the ones where a movement actually took place.

```python
# the default label will be 'idle'
frames["ground_truth"] = "idle"

# iterate over each entry from the ELAN file, select the corresponding frames and label them accordingly:
for idx, ann in annotations.iterrows():
    annotated_frames = (frames.index >= ann["start"]) & (frames.index <= ann["end"])
    frames.loc[annotated_frames, "ground_truth"] = ann["label"]

frames[["left_wrist_y", "right_wrist_y", "ground_truth"]]
```

Now we can temporarily save the frames together with the ground_truth so that this process does not have to be repeated every time. This is especially useful because extracting the movements from the video takes a lot of time.

```python
save_frames = frames.copy()

save_frames.index = save_frames.index.astype('int64') // 1_000_000  # convert index back to integers, so it looks nicer in CSV

# save all data with the new ground_truth column into a new file:
save_frames.to_csv("demo_data/csv_results/demo_video_rotate_frames_with_ground_truth.csv", index=True)

# you can also just save the new ground_truth column into a new file:
save_frames["ground_truth"].to_csv("demo_data/csv_results/demo_video_rotate_only_ground_truth.csv", index=True)
```
