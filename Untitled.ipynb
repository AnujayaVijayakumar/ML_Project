```python
import cv2

cap = cv2.VideoCapture(0, cv2.CAP_V4L2)

while True:
    ok, frame = cap.read()
    if not ok:
        print("no frame")
        break

    cv2.imshow("cam", frame)

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
