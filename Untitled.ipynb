```python
import matplotlib.pyplot as plt
import pandas as pd

videos_to_check = [
    ("ahad_swipe_right", "data/ahad_data/ahad_swipe_right.mp4", "data/ahad_data/ahad_swipe_right_annotation.txt", True),
    ("anu_swipe_right", "data/anujaya_data/anu_swipe_right.mp4", "data/anujaya_data/anu_swipe_right_annotation.txt", True),
]

def inspect_swipe_right(name, video_path, ann_path, flip_image=True):
    df = create_and_annotate_dataframe_from_video(video_path, ann_path, flip_image=flip_image)

    # normalize labels just in case
    df["ground_truth"] = (
        df["ground_truth"].astype(str)
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
    )

    # focus only on annotated gesture frames
    gesture_df = df[df["ground_truth"] == "swipe_right"].copy()
    if len(gesture_df) == 0:
        print(f"\n{name}: no swipe_right frames found")
        return None

    left_dx = gesture_df["left_wrist_x"].iloc[-1] - gesture_df["left_wrist_x"].iloc[0]
    right_dx = gesture_df["right_wrist_x"].iloc[-1] - gesture_df["right_wrist_x"].iloc[0]

    print(f"\n{name}")
    print(f"frames in gesture: {len(gesture_df)}")
    print(f"left_wrist_x change : {left_dx:.5f}")
    print(f"right_wrist_x change: {right_dx:.5f}")

    plt.figure(figsize=(8, 4))
    plt.plot(gesture_df.index.total_seconds(), gesture_df["left_wrist_x"], label="left_wrist_x")
    plt.plot(gesture_df.index.total_seconds(), gesture_df["right_wrist_x"], label="right_wrist_x")
    plt.title(name)
    plt.xlabel("time (s)")
    plt.ylabel("x")
    plt.legend()
    plt.show()

    return {
        "name": name,
        "left_dx": left_dx,
        "right_dx": right_dx,
        "n_frames": len(gesture_df),
    }

results = []
for item in videos_to_check:
    out = inspect_swipe_right(*item)
    if out is not None:
        results.append(out)

results
```
